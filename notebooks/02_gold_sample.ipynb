# Entrega 02: muestra del Gold

100 filas aleatorias de `data/03_gold/gold.parquet` (semilla 5) → `docs/entregas/02_gold_sample.csv`. NHANES es de dominio público.

Requiere haber ejecutado antes el pipeline desde la raíz: `.venv/Scripts/python -m src.main`. No importa nada de `src/`.

```python
# Imports
from pathlib import Path

import pandas as pd
```

```python
# Paths
ROOT = Path.cwd().parent
GOLD_TABLE = ROOT / "data" / "03_gold" / "gold.parquet"
SAMPLE = ROOT / "docs" / "entregas" / "02_gold_sample.csv"

# Sample
SAMPLE_SIZE = 100
SEED = 5
```

```python
gold = pd.read_parquet(GOLD_TABLE)
sample = gold.sample(n=SAMPLE_SIZE, random_state=SEED).sort_values("seqn")

sample.to_csv(SAMPLE, index=False)
print(f"{len(sample)} rows, {int(sample['death'].sum())} deaths, {sample.shape[1]} columns → {SAMPLE.relative_to(ROOT)}")
```
